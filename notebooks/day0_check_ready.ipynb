```python
# 3. One cheap call through the Agents SDK (the stack used in every lab)
from agents import Agent, Runner
MODEL = "gpt-4o"
agent = Agent(name="check", model=MODEL, instructions="Reply with exactly: ready")
result = await Runner.run(agent, "Are you ready?")
print("model said:", result.final_output.strip())
assert "ready" in result.final_output.lower()
```

```python
# 4. Embeddings + an in-process vector store (Day 1 RAG lab)
from openai import OpenAI
client = OpenAI()
emb = client.embeddings.create(model="text-embedding-3-small", input=["brie is a soft cheese"]).data[0].embedding
import chromadb
col = chromadb.Client().create_collection("check")
col.add(ids=["1"], embeddings=[emb], documents=["brie is a soft cheese"])
print("embedding dim", len(emb), "| chroma count", col.count())
print("✅ ready")
```
